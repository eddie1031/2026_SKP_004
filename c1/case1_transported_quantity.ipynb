# 월별 물동량 장기 성장/계절성/환적 구조 파악

부산항만공사에서 제공하고 있는 물동량 예측 기초 데이터를 기반으로 연간 성장률, 월별 계절지수, 환적비중의 장기 변화를 계산했습니다.


## 계절지수

계절지수는 특정 월의 물동량이 평소 수준과 비교했을 때 상대적으로 많은 달인지, 적은 달인지를 숫자로 나타낸 값입니다. 부산항의 월별 물동량을 살펴보면 모든 달에 비슷한 양의 화물이 처리되는 것은 아닙니다. 연초나 연말, 명절 전후, 제조업의 생산 일정, 수출입 기업의 재고 조정, 글로벌 해운 일정 등 여러 요인에 따라 어떤 달에는 물동량이 반복적으로 증가하고, 어떤 달에는 반복적으로 감소할 수 있습니다. 이렇게 매년 비슷한 시기에 반복해서 나타나는 월별 변동의 특징을 계절성이라고 하며, 그 계절성이 어느 정도인지를 숫자로 표현한 것이 계절지수입니다.

예를 들어 부산항의 최근 여러 해 데이터를 살펴보았더니 8월의 물동량이 매년 다른 달보다 조금 많은 경향이 있다고 가정해 보겠습니다. 단순히 8월 물동량의 평균만 계산하면 8월이 실제로 특별히 많은 달인지 판단하기 어렵습니다. 부산항 전체 물동량 자체가 장기적으로 성장하고 있을 수도 있기 때문입니다. 10년 전의 월평균 물동량과 최근의 월평균 물동량 수준이 다르다면 단순한 물동량 숫자만 비교해서는 장기 성장 때문에 커진 것인지, 8월이라는 계절적 특성 때문에 커진 것인지 구분하기 어렵습니다. 계절지수는 이런 문제를 줄이기 위해 각 월의 물동량을 그 시기의 일반적인 수준과 비교하여 상대적인 크기로 바꾸어 봅니다.

계절지수를 1을 기준으로 표현한다고 생각하면 이해하기 쉽습니다. 계절지수가 1.00인 달은 일반적인 월평균 수준과 거의 비슷한 달이라는 뜻입니다. 계절지수가 1.10이라면 그 달의 물동량이 평소보다 약 10% 높은 경향이 있다는 뜻이고, 0.92라면 평소보다 약 8% 낮은 경향이 있다는 뜻입니다. 어떤 분석에서는 이를 100을 기준으로 표현하기도 합니다. 이 경우 110은 평균보다 약 10% 높은 달, 92는 평균보다 약 8% 낮은 달을 의미합니다. 표현 방식만 다를 뿐 의미는 같습니다.

예를 들어 최근 여러 해의 부산항 월별 계절지수가 1월 0.91, 2월 0.94, 3월 1.03, 4월 1.02, 5월 1.01, 6월 1.04, 7월 1.05, 8월 1.07처럼 나타났다고 가정해 보겠습니다. 이 결과를 보고 단순히 "8월 물동량이 가장 많았다"라고 해석하는 것이 아니라, 여러 해를 반복해서 관찰했을 때 8월에는 부산항의 평상시 월간 수준보다 물동량이 약 7% 높은 경향이 있었다라고 해석할 수 있습니다. 반대로 1월의 계절지수가 0.91이라면 여러 해 동안 1월은 일반적인 월보다 약 9% 낮은 물동량을 보이는 경향이 있었다는 뜻입니다.

여기서 중요한 점은 계절지수가 특정 연도 특정 월의 실적을 설명하는 숫자가 아니라는 것입니다. 예를 들어 2025년 8월에 물동량이 많았다고 해서 바로 "8월은 성수기다"라고 판단할 수는 없습니다. 그해에만 발생한 특수한 사건 때문일 수도 있기 때문입니다. 반면 2016년부터 2025년까지처럼 여러 개의 완전한 연도를 확인했을 때 8월의 상대적 물동량이 반복해서 높았다면, 그때부터 8월에 어느 정도 계절성이 존재한다고 판단할 근거가 생깁니다.

이 때문에 이 실습에서 최근 10개 완전 연도를 사용하는 것도 의미가 있습니다. 한두 해만 사용하면 코로나19, 글로벌 물류대란, 항만 파업, 경기침체처럼 특정 연도에 발생한 충격이 계절성처럼 보일 가능성이 있습니다. 여러 해를 함께 사용하면 특정 연도에만 발생했던 우연한 변동의 영향은 어느 정도 줄어들고, 매년 반복되는 월별 패턴을 상대적으로 더 잘 확인할 수 있습니다. 또한 불완전한 연도, 예를 들어 1월부터 8월까지만 데이터가 있는 연도를 그대로 포함하면 9~12월 데이터가 존재하지 않기 때문에 월별 비교가 왜곡될 수 있습니다. 그래서 먼저 연도별 월 수를 확인하고 12개월이 모두 존재하는 완전 연도만 선택하는 과정이 필요합니다.

계절지수를 구하는 목적은 결국 "부산항의 물동량에는 매년 반복되는 월별 패턴이 존재하는가?"라는 질문에 답하기 위한 것입니다. 연간 총량과 전년 대비 성장률은 부산항 물동량이 장기적으로 성장했는지 감소했는지를 보여주지만, 그 안에서 1월부터 12월까지 물동량이 어떻게 반복적으로 움직이는지는 알려주지 않습니다. 반대로 계절지수를 계산하면 "어떤 달이 상대적으로 물동량이 집중되는가", "어떤 달이 상대적으로 한산한가", "이러한 차이가 매년 어느 정도 반복되는가"를 확인할 수 있습니다.

따라서 장기 성장과 계절성은 서로 다른 개념으로 구분해서 보아야 합니다. 부산항의 연간 물동량이 10년에 걸쳐 지속적으로 증가했다면 그것은 장기적인 성장 추세입니다. 그런데 그 성장 과정에서도 매년 1~2월에는 상대적으로 물동량이 낮고 7~9월에는 상대적으로 높아지는 현상이 반복된다면 그것이 계절성입니다. 쉽게 말해 장기 성장은 "시간이 지나면서 전체 규모가 커지고 있는가"를 보는 것이고, 계절지수는 "한 해 안에서 어느 달이 평소보다 많거나 적은가"를 보는 것입니다.

# 데이터 정제

1. 분석에 필요한 형태로 데이터를 준비합니다.

```python
# 연월을 날짜형 데이터로 변경
df['연월'] = pd.to_datetime(
    df['연월'],
    format='%Y-%m',
    errors='coerce'
)

df.dtypes
```

- 연도별 데이터의 완전성 확인

여기서의 완전성은 다른 데이터와 공정하게 비교할 수 있을 정도로 필요한 관측값이 모두 갖추어져있는가를 의미합니다.

```python
# 완전연도(연도데이터의 완전성)
completeness = df.groupby('연도').agg(
    월수=('월', 'nunique'),
    구분수=('수출입구분설명', 'nunique'),
    행수=('물동량', 'size')
)

completeness
```

- 완전연도가 되기 위한 조건: 당해연도에 12개월이 정확히 포함되어있고, 4개의 물동량 구분과, 예상한 48개의 값(매월 4개 수출구분코드에 해당하는 데이터)가 포함되어있는지

```python
has_12_month = completeness['월수'] == 12
has_4_categories = completeness['구분수'] == 4
has_48_rows = completeness['행수'] == 48
```

```python
complete_years = completeness[
    has_12_month
    & has_4_categories
    & has_48_rows
]

complete_years
```

2. 데이터 품질 확인

```python
expected = df.groupby(['연도', '월'])['수출입구분설명'].nunique()

expected
```

```python
quality_info = pd.Series({
    '연월 변환 실패 갯수': df['연월'].isna().sum(),
    '완전 중복 행': df.duplicated().sum(),
    '데이터가 48행이아닌 행 수': expected.ne(4).sum(),
    '완전 연도 수': len(complete_years),
    '불완전 연도 수': completeness.index.difference(complete_years.index).size,
})

quality_info
```

2025년에는 2개월 분량의 데이터만 수집된 것을 확인할 수 있습니다.

그렇다면 2025년과 같이 일부 월만 있는 연도를 연감 증감률 계산에서 제외해야하는데, 어떻게 할 것인지 결정해야 합니다.

```python
annual = df.groupby('연도').agg(
    총물동량=('물동량', 'sum')
)

annual
```

```python
annual['완전연도'] = annual.index.isin(complete_years.index)

annual
```

```python
annual_complete = annual.loc[complete_years.index].copy()
annual_complete['전년대비증감률'] = annual_complete['총물동량'].pct_change()

annual_complete.tail(10)
```

만일 2025년의 데이터가 여기 포함되어있었다면 2025년은 1월,2월에 해당하는 8개의 행만 있기 떄문에 연간 합계를 '24년 기준의 48행과 비교하면 큰 감소처럼 보였을 것입니다. 그렇게 되면 관측 기간의 차이로 인한 증감률이 마치 수요 충격처럼 보일 여지가 있기 때문에, 완전성이 있는 데이터만 포함하여 분석할 수 있도록 한 차례 정제한 것입니다.

# 데이터 분석

- 완전 연도의 총물동량과 전년 대비 증감률에서 최고/최저 성장연도는 언제인가?

```python
growth = annual_complete.dropna(subset=['전년대비증감률'])
min_max = pd.DataFrame({
    '최고 성장': growth.loc[growth['전년대비증감률'].idxmax()],
    '최저 성장': growth.loc[growth['전년대비증감률'].idxmin()],
})

min_max
```

```python
# min_max 전치
min_max = min_max.T

min_max
```

```python
min_max['연도'] = [growth['전년대비증감률'].idxmax(), growth['전년대비증감률'].idxmin()]

min_max[['연도', '총물동량', '전년대비증감률']]
```

완전 연도 가운데 2001년 증가율은 약 26.5%로 가장 높고 2009년은 약 -10.9%로 가장 낮습니다. 데이터상 최근에는 2022년 약 -2.8% 이후 2023년 약 4.9%, 2024년 약 5.4% 증가했습니다.

극단 연도를 찾는 목적은 원인을 자동으로 확정하는 것이 아니라 외부 사건과 집계 변화 여부를 조사할 후보를 정하는 데 있습니다. 전년 대비 증감률은 기준 연도의 크기에 민감하므로 총물동량도 함께 제시했습니다. 한 해의 반등을 장기 성장 추세로 부르려면 이동기간 평균이나 여러 해의 연평균 증가율을 추가로 확인해야 합니다.

- 2015~2024년 계절지수와 10년단위(2001, 2010, 2020, 2024) 환적비중의 의미

```python
# 최근 10년의 월별 물동량 추출
recent = df[df['연도'].between(2015, 2024)].groupby(['연도', '월'])['물동량'].sum().rename('월별합계').reset_index()

recent
```

```python
# 각 연도별 월평균 계산
recent['해당연도월평균'] = recent.groupby('연도')['월별합계'].transform('mean')

# 계절지수 계산
# 계절지수 = 해당 월의 물동량 / 해당 연도의 월평균 물동량
# 계절지수는 물동량 크기가 아니라 당해년도의 평소 수준과 비교하여 특정 달에 상대적으로 높았는지 낮았는지를 나타냅니다.
recent['계절지수'] = recent['월별합계'] / recent['해당연도월평균']

recent
```

```python
# 모든 연도에 대한 월별 집수 재집계(aggregation)
seasonality = recent.groupby('월')['계절지수'].agg(['mean', 'median', 'std'])

seasonality
```

현재 데이터에서는 1.039가 가장 높은 중앙값입니다. 평균은 이 값때문에 올라갈 수 있지만 중앙값은 상대적으로 영향을 덜 받습니다. 따라서 mean과 median을 같이 보면, 특정 몇 년 때문에 평균이 올라간것인지 아니면 대부분의 연도에서 실제로 높은 월인지 판단하는 근거가 됩니다.

- 표준편차
우리 데이터에서 표준편차는 당해 월의 계절성이 연도마다 얼마나 일정했는가를 판단하는 값으로 이해하면 됩니다. 표준편차가 작게나오면 해당 월은 물동량 패턴이 비교적 안정적으로 반복된다고 이해할 수 있고, 표준편차가 크면, 특정 달이 무조건 높은 달이다 라고 단정짓기 어려운것입니다.



특정 달이 높게 집계되었다고 해서, 휴가철 때문에 높다, 특정 수출품 때문에 높다 라고 단정짓는 것은 불가능합니다. 이러한 집계 내용은 단순히 언제 반복적으로 높고 낮은지를 발견하는 분석입니다.

이 분석을 기반으로 왜 이런 데이터가 나왔는지에 대한 합당한 자료는 추가적으로 조사하여 분석해야 하는것입니다.

```python
trade = df.pivot_table(
    index='연도',
    columns='수출입구분설명',
    values='물동량',
    aggfunc='sum',
)

trade
```

```python
trade['환적비중'] = trade.filter(like='환적').sum(axis=1) / trade.sum(axis=1)

trade
```

```python
trade.loc[[2001, 2010, 2020, 2024]]
```

최근 10년의 계절지수 평균은 3월 약 1.039로 가장 높고 2월 약 0.925로 가장 낮습니다. 즉 각 해의 평균적인 달과 비교할 때 3월은 약 3.9% 높고 2월은 약 7.5% 낮은 경향을 알 수 있습니다.

환적비중은 2001년 약 36.5%에서 2024년 약 55.3%로 높아져 부산항의 국제 환적 허브 역할이 강화되었다고 볼 수 있습니다.
